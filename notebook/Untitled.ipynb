```python
import matplotlib.pyplot as plt
import modin.pandas as pd
import seaborn as sns

plt.figure(figsize=(20,15))

# blue
baseline = pd.read_csv("/Users/yamiteru/.isitfast/results/baseline.js-$main-main.csv")
# orange
first = pd.read_csv("/Users/yamiteru/.isitfast/results/array.js-$unsub-main.csv")
# green
second = pd.read_csv("/Users/yamiteru/.isitfast/results/set.js-$unsub-main.csv")

baseline = baseline.groupby(by="run").rolling(20).median()
first = first.groupby(by="run").rolling(20).median()
second = second.groupby(by="run").rolling(20).median()

sns.lineplot(x="iteration", y="ram", data=baseline)
sns.lineplot(x="iteration", y="ram", data=first)
sns.lineplot(x="iteration", y="ram", data=second)
```
